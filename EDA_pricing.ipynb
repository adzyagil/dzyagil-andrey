# EDA анализ для полученного решения

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set(style="whitegrid")
np.random.seed(322)

train = pd.read_csv("train.csv")
train["dt"] = pd.to_datetime(train["dt"])

print(train.shape)
display(train.head())
display(train.describe().T)
print(train.info())
```

```python
test = pd.read_csv("train.csv")
test["dt"] = pd.to_datetime(train["dt"])

print(test.shape)
display(test.head())
display(test.describe().T)
print(test.info())
```

Данные хорошо структурированы, не содержат пропусков и имеют выраженную
временную и категориальную структуру.

Типы данных в датасете:
- 7 вещественных признаков (цены, погода, количество магазинов),
- 11 целочисленных (категории, календарные и бинарные флаги),
- 1 временной признак (dt).

Особенности наблюдений:
- Период наблюдений: с 28 марта 2024 по 26 мая 2024, то есть около 2 месяцев ежедневных данных.
- Каждое наблюдение соответствует паре (product_id, день).
- случайный shuffle недопустимым — корректное разделение должно быть временным.

Анализ целевых переменных:
- Средние значения: price_p05: ~1.02, price_p95: ~1.12
- Разброс цен умеренный, но присутствуют выбросы (максимумы выше 3.5)
- Минимальные значения доходят до 0, что важно учитывать при лог-преобразованиях.
- Стандартные отклонения у p05 и p95 сопоставимы, но сами интервалы не симметричны относительно центра.
  
Ключевые сложности задачи:
- необходимость предсказывать не точку, а интервал (что может условжнить задачу с подбором адекватного интервала)
- асимметрия распределений
- зависимость от истории конкретного товара.

Это делает естественным использование:
- rolling-признаков,
- декомпозиции интервала,
- квантильной регрессии.

```python
fig, ax = plt.subplots(1, 2, figsize=(14, 5))
sns.histplot(train["price_p05"], bins=50, ax=ax[0])
sns.histplot(train["price_p95"], bins=50, ax=ax[1])
ax[0].set_title("Распределение price_p05")
ax[1].set_title("Распределение price_p95")
plt.show()
```

Как уже упоминалось ранне, интервалы не симметричны отсноительно центьра. Цены имеют выраженную асимметрию и длинные хвосты.
Это делает лог-преобразования естественным выбором.

```python
train["width"] = train["price_p95"] - train["price_p05"]

# берем топ товаров по числу наблюдений
top_products = train["product_id"].value_counts().head(10).index

plt.figure(figsize=(10, 4))
train[train.product_id.isin(top_products)].boxplot(
    column="width",
    by="product_id",
    grid=False
)
plt.title("Распределение ширины интервала по товарам")
plt.suptitle("")
plt.xlabel("product_id")
plt.ylabel("p95 - p05")
plt.show()
```

Распределение ширины интервала существенно различается
между товарами.

Это подтверждает, что неопределенность цены является
товар-специфичной характеристикой и требует учета
истории конкретного продукта.


```python
# вычисляем центр и дельты
train["center"] = (train["price_p05"] + train["price_p95"]) / 2
train["lower_delta"] = train["center"] - train["price_p05"]
train["upper_delta"] = train["price_p95"] - train["center"]

# визуализируем распределения
fig, ax = plt.subplots(1, 2, figsize=(14,5))
sns.histplot(train["lower_delta"], bins=50, ax=ax[0])
sns.histplot(train["upper_delta"], bins=50, ax=ax[1])
ax[0].set_title("Lower delta distribution")
ax[1].set_title("Upper delta distribution")
plt.show()

```

Вообще на данном графике распределение нижней и верхней границ идентичное и возможно испольщование симметричных интревалов сработает. Но я попробовал оба варианта и скор при несимметричном интервале отработал немоного лучше

```python
example_pid = train["product_id"].iloc[0]
tmp = train[train.product_id == example_pid].sort_values("dt")

plt.figure(figsize=(14,5))
plt.plot(tmp["dt"], tmp["center"], label="center")
plt.fill_between(tmp["dt"], tmp["price_p05"], tmp["price_p95"], alpha=0.3)
plt.title(f"Price dynamics for product {example_pid}")
plt.legend()
plt.show()
```

Для одного и того же товара цены:
- обладают локальной автокорреляцией
- имеют относительно устойчивые паттерны во времени
- Хотя 22 мая произошел достаточно большой выброс, но во все остальные даты динамика достаточно устоцчивая

Это дает нам идею о возможной пользе от использования rolling-фичей.


```python
train = train.sort_values(["product_id", "dt"])
train["c_mean_7"] = train.groupby("product_id")["center"].shift(1).rolling(7).mean()

sns.scatterplot(
    data=train.sample(5000, random_state=322),
    x="c_mean_7",
    y="center",
    alpha=0.3
)
plt.show()
```

Rolling-средние центра интервала имеют достточно хорошую взаимосвязь
с текущими значениями, особенно для частых товаров.

Это делает их сильным признаков.

У нас нет откровенно редких категорих и какого-то прям серьезного дисбаланса

```python

eda_cols = [
    "price_p05", "price_p95",
    "n_stores", "precpt",
    "avg_temperature", "avg_humidity", "avg_wind_level",
    "holiday_flag", "activity_flag"
]

corr = train[eda_cols].corr()

plt.figure(figsize=(8, 6))
sns.heatmap(
    corr,
    cmap="coolwarm",
    center=0,
    annot=True,
    fmt=".2f",
    square=True,
    cbar_kws={"shrink": 0.8}
)
plt.title("Корреляционная матрица основных числовых признаков")
plt.show()
```

price_p05 и price_p95 сильно коррелируют
Погодные признаки почти не коррелируют с ценами

```python
# Относительная ширина интервала и эффект малых цен
train["width"] = train.price_p95 - train.price_p05
train["rel_width"] = train["width"] / (train["price_p05"] + 1e-6)

train["price_bin"] = pd.qcut(train["price_p05"], 10)

train.groupby("price_bin")["rel_width"].mean()
```

Мы рассмотрели относительную ширину интервала:
и сгруппировали наблюдения по бинам значения `price_p05`.

Результаты показали резкий всплеск относительной ширины
для товаров с очень низкой ценой.

Это связано с тем, что при `price_p05 около 0`
даже небольшой абсолютный интервал
приводит к взрывному росту относительной метрики.

```python
# проверим активность и праздники
train.groupby(["holiday_flag", "activity_flag"])["width"].mean()
```

Сравнение флагов показало, что признак активности
оказывает более выраженное влияние на ширину интервала,
чем сам факт праздничного дня.


```python
# глянем на лаги и память
for lag in [1, 3, 7, 14,28]:
    train[f"lag_width_{lag}"] = train.groupby("product_id")["width"].shift(lag)

train[[f"lag_width_{l}" for l in [1,3,7,14,28]]].corr()["lag_width_1"]
```

Вполне логично, что болеее свежжие лаги лучше всего предсказывают будущее в карткочроке, а на долгосроке не так зороши

Категориальные признаки таким образом

Небольшие рассуждения про нашу метрику

IoU поощряет:
- правильную форму интервала
- асимметрию
- адаптацию ширины под неопределённость

Поэтому точечная регрессия + фиксированная ширина
дает слабый результат.


### Логика финального решения
Итоговое решение строится на следующих принципах:

1. Декомпозиция интервала:
   - центр
   - нижнее и верхнее отклонения

2. Лог-преобразования для стабилизации распределений

3. Rolling-признаки для учета временной динамики

4. Квантильная регрессия:
   - медиана для центра
   - асимметричные квантили для границ

5. Усреднение с продуктовыми статистиками
   для редких товаров
